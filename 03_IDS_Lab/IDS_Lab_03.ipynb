```python
import pandas as pd
```

```python
df = pd.read_csv('sales.csv')
print("Sales Dataset Loaded Successfully!\n")
print(df.head())
```

```python
print("Summary Statistics:")
print(df.describe())
```

```python
print("\nMean of Columns:")
print(df.mean(numeric_only=True))
```

```python
print("\n Median of Columns:")
print(df.median(numeric_only=True))
```

```python
print("\n Maximum Values:")
print(df.max(numeric_only=True))

print("\n Minimum Values:")
print(df.min(numeric_only=True))
```

```python
print("\nCount of Non-null Values:")
print(df.count())
```

```python
print("\n Standard Deviation of Columns:")
print(df.std(numeric_only=True))
```

```python
print("\nCorrelation Between Columns:")
print(df.corr(numeric_only=True))
```
