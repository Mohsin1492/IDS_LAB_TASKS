```python
import pandas as pd
```

```python
df = pd.read_csv('Lab5dataset.csv')
```

```python
df.groupby('Product')['Total_Sales'].sum().plot(
    kind='bar', title='Total Sales by Product'
)
```

```python
df.plot(x='Product', y='Total_Sales', kind='line',
        marker='o', title='Sales Trend by Product')
```

```python
df.groupby('Category')['Total_Sales'].sum().plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90,
    title='Sales Percentage by Category',
    ylabel=''
)
```

```python
import matplotlib.pyplot as plt
```

```python
# Prepare aggregated data
product_sales = df.groupby('Product')['Total_Sales'].sum()
category_sales = df.groupby('Category')['Total_Sales'].sum()
```

```python
plt.figure(figsize=(8, 5))
plt.bar(product_sales.index, product_sales.values)
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(product_sales.index, product_sales.values, marker='o')
plt.title("Sales Trend by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 7))
plt.pie(
    category_sales.values,
    labels=category_sales.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Sales Percentage by Category")
plt.show()
```

```python

# 1. Which product has the highest sales?
product_sales = df.groupby('Product')['Total_Sales'].sum()
print("Highest-selling product:", product_sales.idxmax())
print("Total sales:", product_sales.max())

# 2. Which category contributes the most?
category_sales = df.groupby('Category')['Total_Sales'].sum()
print("Top category:", category_sales.idxmax())
print("Category sales:", category_sales.max())

# 3. Is there any trend in sales?
print("Sales by product:")
print(product_sales)

# 4. What does the histogram show about distribution?
print(df['Total_Sales'].describe())

# 5. Is there a relationship between price and sales?
correlation = df['Unit_Price'].corr(df['Total_Sales'])
print("Price-sales correlation:", correlation)
```
