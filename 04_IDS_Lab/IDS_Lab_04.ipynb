```python
import pandas as pd
```

```python
df = pd.read_csv("sales_data.csv")
```

```python
print(df.head())
```

```python
category_sales = df.groupby('Category')['Total_Sales'].sum()
print("Total Sales per Category:\n", category_sales)
```

```python
avg_price = df.groupby('Product')['Unit_Price'].mean()
print("\nAverage Unit Price per Product:\n", avg_price)
```

```python
df['Discounted_Sales'] = df['Total_Sales'].apply(lambda x: x * 0.9)
print("\nDataFrame with Discounted Sales:\n", df[['Product', 'Total_Sales', 'Discounted_Sales']])
```

```python
sorted_sales = df.sort_values(by='Total_Sales', ascending=False)
print("\nData sorted by Total Sales:\n", sorted_sales)
```
